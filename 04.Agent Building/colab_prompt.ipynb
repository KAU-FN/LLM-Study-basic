```python
import os
from huggingface_hub import InferenceClient
client = InferenceClient(model="moonshotai/Kimi-K2.5")
```

```python
output = client.chat.completions.create(
    messages=[
        {"role": "user", "content": "파리의 수도는?"},
    ],
    stream=False,
    max_tokens=1024,
    extra_body={'thinking': {'type': 'disabled'}},
)
print(output.choices[0].message.content)
```

```python
SYSTEM_PROMPT = """다음 질문에 최선을 다해 답하세요.

당신은 다음 도구들을 사용할 수 있습니다:

get_weather: 특정 지역의 현재 날씨를 가져옵니다.

도구를 사용하려면 JSON 형식을 작성해야 합니다.
구체적으로, 이 JSON에는 `action` 키(사용할 도구의 이름)와 `action_input` 키(도구에 전달할 입력값)가 포함되어야 합니다.

"action" 필드에 들어갈 수 있는 값은 오직 다음과 같습니다:
get_weather: 특정 지역의 현재 날씨를 가져옵니다, 인자(args): {"location": {"type": "string"}}
사용 예시:

{{
  "action": "get_weather",
  "action_input": {"location": "뉴욕"}
}}


항상 다음 구조를 엄격하게 따르세요:

Question: 당신이 대답해야 할 입력 질문
Thought: 어떤 행동을 취할지 항상 생각해야 합니다. 한 번에 단 하나의 행동(Action)만 다음 형식으로 취하세요:
Action:

$JSON_BLOB (마크다운 코드 블록 내부에 작성)

Observation: 행동의 결과. 이 관찰(Observation) 결과는 고유하고 완전하며 신뢰할 수 있는 정보의 원천입니다.
(이 Thought/Action/Observation 과정은 N번 반복될 수 있으며, 필요하다면 여러 단계를 거쳐야 합니다. $JSON_BLOB은 반드시 마크다운 형식으로 작성되어야 하며, 한 번에 오직 하나의 행동만 사용해야 합니다.)

출력을 마무리할 때는 항상 다음 형식을 사용해야 합니다:

Thought: 이제 최종 답변을 알겠습니다.
Final Answer: 원래 입력된 질문에 대한 최종 답변

이제 시작하세요! 최종적인 답변을 제공할 때는 항상 `최종 답변:`라는 문자열을 정확히 사용해야 함을 잊지 마세요. """
```

```python
messages = [
    {"role": "system", "content": SYSTEM_PROMPT},
    {"role": "user", "content": "현재 서울의 날씨가 어때?"},
]

print(messages)
```

```python
output = client.chat.completions.create(
    messages=messages,
    stream=False,
    max_tokens=200,
    extra_body={'thinking': {'type': 'disabled'}},
)
print(output.choices[0].message.content)
```

# **어 케 했 노?**

진짜 LLM이 서울의 날씨를 가져왔는지 확인해보자

```python
output = client.chat.completions.create(
    messages=messages,
    max_tokens=150,
    stop=["Observation:"], # 이 인자를 넣으면 관찰단계 즉, 날씨 확인 직전에 멈춘다
    extra_body={'thinking': {'type': 'disabled'}},
)

print(output.choices[0].message.content)
```

이제 get_weahter 함수를 만들어 제공해보자

```python
def get_weather(location):
    return f"{location}의 날씨는 봄부터 벌써 더워 죽을것 같읍니다."

get_weather('런던')
```

```python
messages=[
    {"role": "system", "content": SYSTEM_PROMPT},
    {"role": "user", "content": "서울의 날씨는 어때???"},
    {"role": "assistant", "content": output.choices[0].message.content + "Observation:\n" + get_weather('서울')},
]

output = client.chat.completions.create(
    messages=messages,
    stream=False,
    max_tokens=1000,
    extra_body={'thinking': {'type': 'disabled'}},
)
print(output.choices[0].message.content)
```
