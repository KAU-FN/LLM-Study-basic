Action을 Json이나 Text가 아닌 코드로 정의하는 특징이 있음

```python
try:
            # 에이전트 객체 선언
            agent_instance = Agent(
                model=HF_MODEL,             # 사용할 모델
                provider=DEFAULT_PROVIDER,  # Hugging Face, Antropic 등등
                api_key=HF_TOKEN,           # 인증 토큰

                servers=[
                    {
                        # sse 사용하고 싶으면 "sse" 현재는 파이썬 입출력
                        "type": "stdio",

                        # 서버 실행 스크립트 (도커랑 비슷하네..)
                        "command": "python",
                        "args": ["mcp_server.py"],

                        # 작업 디렉토리 설정
                        "cwd": ".",

                        # 환경변수 설정
                        "env": {"HF_TOKEN": HF_TOKEN} if HF_TOKEN else {},
                    }
                ],
            )
            print("✅ Agent instance created successfully")

            print("🔧 Loading tools...")
            # 서버 구동되고 Agent에게 사용가능한 Tool 로드
            await agent_instance.load_tools()
            print("✅ Tools loaded successfully")

        except Exception as e:
            print(f"❌ Error creating/loading agent: {str(e)}")
            agent_instance = None
```
